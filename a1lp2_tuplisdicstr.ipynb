```python
# ----- Tuple Operations -----
my_tuple = (10, 20, 30, 40, 50)
print("Original Tuple:", my_tuple)

# Accessing elements
print("First element:", my_tuple[0])
print("Slice (1:4):", my_tuple[1:4])

# ----- List Operations -----
my_list = [1, 2, 3, 4, 5]
print("\nOriginal List:", my_list)

my_list.append(6)
print("After append:", my_list)

my_list.insert(2, 10)
print("After insert at index 2:", my_list)

my_list.remove(4)
print("After removing 4:", my_list)

my_list[0] = 99
print("After updating first element:", my_list)

# ----- Dictionary Operations -----
my_dict = {"name": "Alex", "age": 21, "city": "Mumbai"}
print("\nOriginal Dictionary:", my_dict)

# Accessing value
print("Name:", my_dict["name"])

# Adding new key-value pair
my_dict["college"] = "XYZ Institute"
print("After adding new key:", my_dict)

# Updating value
my_dict["age"] = 22
print("After updating age:", my_dict)

# Deleting a key
del my_dict["city"]
print("After deleting 'city':", my_dict)
```

```python
# Taking input string
text = "Hello Python Programming"

print("Original String:", text)
print("Length of string:", len(text))

# Accessing characters
print("First character:", text[0])
print("Last character:", text[-1])

# Uppercase & Lowercase
print("Uppercase:", text.upper())
print("Lowercase:", text.lower())

# Reversing string
print("Reversed String:", text[::-1])

# Splitting string
words = text.split()
print("List of words:", words)

# Replacing substring
print("Replace 'Python' with 'Data':", text.replace("Python", "Data"))

# Checking types
print("Is alphabetic?", text.isalpha())
print("Is numeric?", text.isnumeric())
```
